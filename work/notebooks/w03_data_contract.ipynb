## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

This data cannot reliably tell us about clients with equal historical coverage because the client histories are unbalanced. Early rows may be GSC-only, so the absence of GA4 data does not necessarily mean that GA4 access was unavailable. Overlapping data windows can also make comparisons across clients misleading. Therefore, conclusions should be treated as observed patterns and decision-support signals, not definitive causal claims.
