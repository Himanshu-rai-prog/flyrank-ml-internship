## 8. 5-Minute Demo Outline

### 1. Question — 45 seconds
Can a machine-learning ranking model identify content pages showing signals of decline more effectively than a simple transparent refresh rule?

The practical problem is prioritization: when many content pages exist, a team needs a repeatable way to identify which pages deserve refresh review first.

### 2. Method — 1 minute 30 seconds
I worked with 30,000 anonymized content records containing search, engagement, content, and lifecycle features.

I first created a transparent baseline rule: review pages that had not been updated for at least 181 days and still had at least 500 impressions in the last 90 days.

I then used Logistic Regression to rank pages by predicted probability of decline, while excluding target/trend fields and recent/previous 30-day fields from the model features to reduce leakage risk.

### 3. One Chart — 1 minute
Show the Precision@50 comparison chart from the research paper.

The baseline measured Precision@50 of 0.52, while the Logistic Regression model measured 0.64 on the held-out test split.

The key message is that the model surfaced more declining pages within the top 50 recommendations on this validation setup.

### 4. Honest Result — 45 seconds
At Precision@100, the baseline measured 0.56 and the model measured 0.66.

These are observed results on this dataset and held-out test split. They are directional decision-support results, not evidence that the same performance will occur for every future client or dataset.

### 5. Recommendation — 1 minute
Use the model as a prioritization aid rather than an automatic refresh decision.

Start review with high-ranked pages, then check whether the page is genuinely outdated, whether the traffic represents useful visibility, and whether a refresh is actually appropriate.

The final decision should remain with human review and business context.
